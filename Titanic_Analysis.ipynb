```python
import pandas as pd
import numpy as np
from google.colab import files
uploaded = files.upload()
#Titanic dataset
titanic = pd.read_csv('train.csv')
# print('shape:', titanic.shape)
# print('columns:', titanic.columns.tolist())
# print('====DATASET INFO ====')
# print(titanic.info)
# print('==== STATISTICS ===')
# print(titanic.describe)


#cleaning the Data
#Fix 1: Fill the Age with the Average
titanic['Age'].fillna(titanic['Age'].mean(), inplace = True)

#Fix 2: due to to the high number of missing record of cabin we can't find the average correct so we drop it the column
titanic.drop("Cabin", axis=1, inplace = True)

#fix 3: fixing the embarked with the only two missing values we get the most occurring value in the column and we fill it with the two missing records
titanic['Embarked'].fillna(titanic['Embarked'].mode()[0], inplace= True)
#print("==== MiSSING VALUES ===")
#print(titanic.isnull().sum())
#print(titanic.shape)
# I want to determing the number of people who survived
print("==== Number of people who survived ===")
survived = titanic['Survived'].value_counts()
print(survived)
print(f"Survived: {survived[1]}")
print(f"Died: {survived[0]}")
print("===Rate of people who are alive ===")
alive = round(titanic['Survived'].mean()* 100, 2)
print(f"{alive}%")
print("=== Rate of people who didn't survived ===")
died = 100 - alive
print(f"{died}%")

#Gender Analysis
print("=== GENDER ANALYSIS ===")
gender = titanic.groupby('Sex')['Survived'].mean()* 100
print(gender)
print("Male Survivde Rate:", round(titanic[titanic['Sex']=='male']["Survived"].mean()*100, 2), "%")
print("Female Survivde Rate:", round(titanic[titanic['Sex']=='female']["Survived"].mean()*100, 2), "%")
# Passengers class analysis
print("====  PASSENGERS CLASS ANALYSIS ===")
pclass = titanic.groupby('Sex')["Survived"].mean()* 100
print(pclass)
print("\nIst Class Survived Rate:", round(titanic[titanic["Pclass"]==1]["Survived"].mean()*100, 2),"%")
print("\n2nd Class Survived Rate:", round(titanic[titanic["Pclass"]== 2]["Survived"].mean()*100, 2), "%")
print("\n3rd Class Survived Rate:", round(titanic[titanic["Pclass"]== 3]["Survived"].mean()*100, 2), "%")

#Age Analysis
print("==== AGE ANALYSIS ===")
print("\nAverage Age:", round(titanic["Age"].mean(),  2))
print('Youngest Passenger:', titanic['Age'].min())
print("Oldest Passenger:", titanic["Age"]. max())

#survied vs non survived
print("==== Survived vs Non Survived ===")
print("\nAverage Age Survived:", round(titanic[titanic["Survived"]==1]["Age"].mean(), 2))
print("Average Age not Survived:", round(titanic[titanic["Survived"]==0]["Age"].mean(), 2))

print("==============================")
print("TITANIC ANALYSIS SUMMARY")
print("==============================")
print(f"Total Passengers: 891")
print(f"Survived: 342 (38.38%)")
print(f"Died: 549 (61.62%)")
print("------------------------------")
print(f"Female survival: 74.20%")
print(f"Male survival: 18.89%")
print("------------------------------")
print(f"1st Class survival: 62.96%")
print(f"2nd Class survival: 47.28%")
print(f"3rd Class survival: 24.24%")
print("------------------------------")
print(f"Youngest passenger: 0.4 yrs")
print(f"Oldest passenger: 80 yrs")
print(f"Avg age survivors: 28.55")
print(f"Avg age non-survivors: 30.42")
print("==============================")
print("Key Finding: Gender and Class")
print("were the biggest survival factors")
print("==============================")





```
